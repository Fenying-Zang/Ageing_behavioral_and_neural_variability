```python
# Make sure you're with IBL environment activated
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
import logging, sys
from one.api import ONE

TAG = '2025_Q3_Zang_et_al_Aging'  

one = ONE(
            mode="remote",
            base_url="https://openalyx.internationalbrainlab.org",
            password="international",
            silent=True,
        )
all_sessions_included =  list(one.alyx.rest('sessions', 'list', tag=TAG))
print(f"Total number of sessions with tag {TAG}: {len(all_sessions_included)}")
```

```python
#To access the lifespan learning dataset, you can use the ONE API to query the Alyx database. 
project_name = 'churchland_learninglifespan'
sessions_lifespan =  list(one.alyx.rest('sessions', 'list', tag=TAG, projects=project_name))
print(f'Number of sessions with project {project_name}: {len(sessions_lifespan)}')
```

```python
# To access the brainwide map dataset, you can similarly query the Alyx database using the ONE API. 
project_name = 'ibl_neuropixel_brainwide_01'
sessions_BWM =  list(one.alyx.rest('sessions', 'list', tag=TAG, projects=project_name))
print(f'Number of sessions with project {project_name}: {len(sessions_BWM)}')
```
